# 6. Visualization dashboard

This notebook regenerates and displays tutorial figures.

```python
import sys
from pathlib import Path

def _find_pkg_dir(start):
 cur = Path(start).resolve()
 for cand in [cur, *cur.parents]:
 for sub in ("source", "experiment"):
 if (cand / sub / "run.py").exists():
 return cand / sub
 return None

PROJECT_ROOT = Path.cwd()
EXPERIMENT_DIR = _find_pkg_dir(PROJECT_ROOT)
if EXPERIMENT_DIR is None:
 raise FileNotFoundError("Could not locate source/run.py (or source/run.py). Run this notebook from the repository root, or `pip install -e source` first.")
PROJECT_ROOT = EXPERIMENT_DIR.parent
RESULTS_DIR = PROJECT_ROOT / "results"
sys.path.insert(0, str(EXPERIMENT_DIR))
print("EXPERIMENT_DIR =", EXPERIMENT_DIR)
print("Experiment code exists:", (EXPERIMENT_DIR / "run.py").exists())
```

```python
from visualize import create_figures
latest_metrics = sorted(RESULTS_DIR.glob("run_*_real_metrics.json"))[-1]
latest_raw = Path(str(latest_metrics).replace("_metrics.json", "_raw.json"))
figures = create_figures(latest_raw, latest_metrics, RESULTS_DIR)
print("Figures:")
for fig in figures:
 print(fig)
```

```python
try:
 from IPython.display import Image, display
 figure_names = [
 "figure_real_retrieval.png",
 "figure_semantic_scores.png",
 "figure_semantic_vs_retrieval.png",
 "figure_evaluator_coverage.png",
 "figure_memory_growth.png",
 "figure_failures.png",
 ]
 for name in figure_names:
 path = RESULTS_DIR / name
 if path.exists():
 print(name)
 display(Image(filename=str(path)))
except Exception as exc:
 print("Display skipped:", exc)
```
